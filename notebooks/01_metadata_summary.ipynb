# Tahoe-100M Metadata Summary

This notebook loads and summarizes the metadata tables from the Tahoe-100M dataset.

```python
import pandas as pd
import json
from pathlib import Path

DATA_DIR = Path("../data/processed")
```

## 1. Drug Metadata

```python
drugs = pd.read_parquet(DATA_DIR / "drug_metadata.parquet")
print(f"Total drugs: {len(drugs)}")
print(f"\nColumns: {list(drugs.columns)}")
drugs.head()
```

```python
# mechanism of action distribution
print("Mechanism of Action (broad) distribution:")
print(drugs['moa-broad'].value_counts().head(15))
```

```python
# look for MEK inhibitors (relevant to benchmark example)
mek_inhibitors = drugs[drugs['moa-fine'].str.contains('MEK', case=False, na=False)]
print(f"\nMEK inhibitors ({len(mek_inhibitors)}):")
mek_inhibitors[['drug', 'moa-fine', 'targets']]
```

## 2. Cell Line Metadata

```python
cell_lines = pd.read_parquet(DATA_DIR / "cell_line_metadata.parquet")
print(f"Total cell line entries: {len(cell_lines)}")
print(f"\nColumns: {list(cell_lines.columns)}")
cell_lines.head()
```

```python
# unique cell lines
unique_cells = cell_lines['cell_name'].nunique()
print(f"Unique cell lines: {unique_cells}")
print(f"\nCell lines: {sorted(cell_lines['cell_name'].unique())}")
```

```python
# organ/tissue distribution
print("Organ distribution:")
print(cell_lines.groupby('cell_name')['Organ'].first().value_counts())
```

```python
# driver gene distribution
print("Driver gene distribution:")
print(cell_lines['Driver_Gene_Symbol'].value_counts().head(20))
```

```python
# KRAS mutations (relevant to benchmark example)
kras_lines = cell_lines[cell_lines['Driver_Gene_Symbol'] == 'KRAS']
print(f"\nCell lines with KRAS mutations ({len(kras_lines)} entries):")
kras_lines[['cell_name', 'Driver_Gene_Symbol', 'Driver_ProtEffect_or_CdnaEffect', 'Organ']].drop_duplicates()
```

## 3. Gene Metadata

```python
genes = pd.read_parquet(DATA_DIR / "gene_metadata.parquet")
print(f"Total genes: {len(genes)}")
print(f"\nColumns: {list(genes.columns)}")
genes.head()
```

```python
# check for our signature genes
with open(DATA_DIR / "gene_signatures.json") as f:
    signatures = json.load(f)

for sig_name, sig_genes in signatures.items():
    found = genes[genes['gene_symbol'].isin(sig_genes)]
    print(f"{sig_name}: {len(found)}/{len(sig_genes)} genes found in dataset")
```

## 4. Summary for Benchmark Episodes

Key statistics relevant for episode generation:

```python
print("=" * 50)
print("TAHOE-100M METADATA SUMMARY")
print("=" * 50)
print(f"\nDrugs: {len(drugs)}")
print(f"  - Unique MOA (broad): {drugs['moa-broad'].nunique()}")
print(f"  - Unique MOA (fine): {drugs['moa-fine'].nunique()}")
print(f"\nCell lines: {cell_lines['cell_name'].nunique()}")
print(f"  - Driver genes represented: {cell_lines['Driver_Gene_Symbol'].nunique()}")
print(f"  - Organs/tissues: {cell_lines.groupby('cell_name')['Organ'].first().nunique()}")
print(f"\nGenes: {len(genes)}")
print(f"\nPotential drug-cell combinations: {len(drugs) * cell_lines['cell_name'].nunique():,}")
```

```python
# find cell lines from the benchmark example
example_cells = ['SW620', 'HT29', 'SW480', 'LoVo', 'Caco2', 'RKO']
print("\nBenchmark example cell lines:")
for cell in example_cells:
    matches = cell_lines[cell_lines['cell_name'].str.contains(cell, case=False, na=False)]
    if len(matches) > 0:
        kras_status = matches[matches['Driver_Gene_Symbol'] == 'KRAS']
        if len(kras_status) > 0:
            mut = kras_status['Driver_ProtEffect_or_CdnaEffect'].iloc[0]
            print(f"  {cell}: KRAS {mut}")
        else:
            print(f"  {cell}: KRAS wildtype (no KRAS mutation listed)")
    else:
        print(f"  {cell}: NOT FOUND in dataset")
```

```python
# find drugs from the benchmark example
example_drugs = ['Trametinib', 'Cobimetinib', 'Binimetinib', 'Selumetinib', 'Alpelisib', 'Venetoclax']
print("\nBenchmark example drugs:")
for drug in example_drugs:
    matches = drugs[drugs['drug'].str.contains(drug, case=False, na=False)]
    if len(matches) > 0:
        moa = matches['moa-fine'].iloc[0]
        print(f"  {drug}: {moa}")
    else:
        print(f"  {drug}: NOT FOUND in dataset")
```
